```python
import numpy as np
import torchvision.transforms as T
from decord import VideoReader, cpu
from PIL import Image
from torchvision.transforms.functional import InterpolationMode

IMAGENET_MEAN = (0.485, 0.456, 0.406)
IMAGENET_STD  = (0.229, 0.224, 0.225)

def build_transform(input_size):
    return T.Compose([
        T.Lambda(lambda img: img.convert("RGB") if img.mode != "RGB" else img),
        T.Resize((input_size, input_size), interpolation=InterpolationMode.BICUBIC),
        T.ToTensor(),
        T.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD)
    ])

def find_closest_aspect_ratio(aspect_ratio, target_ratios, width, height, image_size):
    best_ratio_diff, best_ratio = float("inf"), (1,1)
    area = width * height
    for ratio in target_ratios:
        diff = abs(aspect_ratio - ratio[0]/ratio[1])
        if diff < best_ratio_diff:
            best_ratio_diff, best_ratio = diff, ratio
        elif diff == best_ratio_diff:
            if area > 0.5 * image_size * image_size * ratio[0] * ratio[1]:
                best_ratio = ratio
    return best_ratio

def dynamic_preprocess(image, min_num=1, max_num=6, image_size=448, use_thumbnail=False):
    ow, oh = image.size
    aspect = ow / oh
    target_ratios = sorted(
        set((i,j) for n in range(min_num, max_num+1) for i in range(1,n+1) for j in range(1,n+1) if i*j<=max_num and i*j>=min_num),
        key=lambda x: x[0]*x[1]
    )
    ta = find_closest_aspect_ratio(aspect, target_ratios, ow, oh, image_size)
    tw, th = image_size * ta[0], image_size * ta[1]
    blocks = ta[0] * ta[1]
    resized = image.resize((tw, th))
    processed = []
    for i in range(blocks):
        box = (
            (i % (tw // image_size)) * image_size,
            (i // (tw // image_size)) * image_size,
            ((i % (tw // image_size)) + 1) * image_size,
            ((i // (tw // image_size)) + 1) * image_size
        )
        processed.append(resized.crop(box))
    if use_thumbnail and len(processed) != 1:
        processed.append(image.resize((image_size, image_size)))
    return processed

def get_index(bound, fps, max_frame, first_idx=0, num_segments=32):
    if bound:
        start, end = bound[0], bound[1]
    else:
        start, end = -100000, 100000
    start_idx = max(first_idx, round(start * fps))
    end_idx = min(round(end * fps), max_frame)
    seg_size = float(end_idx - start_idx) / num_segments
    return np.array([
        int(start_idx + seg_size/2 + np.round(seg_size*idx))
        for idx in range(num_segments)
    ])

def load_video(video_path, input_size=448, max_num=1, num_segments=16):
    vr = VideoReader(video_path, ctx=cpu(0), num_threads=1)
    max_frame = len(vr) - 1
    fps = float(vr.get_avg_fps())
    pixel_values_list, num_patches_list = [], []
    transform = build_transform(input_size)
    frame_indices = get_index(None, fps, max_frame, first_idx=0, num_segments=num_segments)
    for idx in frame_indices:
        img = Image.fromarray(vr[idx].asnumpy()).convert("RGB")
        tiles = dynamic_preprocess(img, image_size=input_size, use_thumbnail=True, max_num=max_num)
        pixel_values = torch.stack([transform(tile) for tile in tiles])
        num_patches_list.append(pixel_values.shape[0])
        pixel_values_list.append(pixel_values)
    return torch.cat(pixel_values_list), num_patches_list

# 定义 Prompt
ANOMALY_PROMPT = """你是一个校园监控视频异常行为检测专家。请仔细分析视频，完成以下任务：
1. 判断视频中是否存在异常行为。异常行为包括但不限于：
   - 交通违章：闯红灯、逆行、超速、违停、行人闯入机动车道
   - 危险动作：追逐打闹、翻越护栏或围墙、抛物、打架斗殴
   - 安全违规：未戴安全帽、在禁烟区吸烟、违规使用手机
2. 如果存在异常，请说明：
   - 异常类型（分类）
   - 时间段（前段/中段/后段/全过程）
   - 详细描述
   - 是否为多事件

请严格按照以下 JSON 格式输出（不要输出其他文字）：
{
    "has_anomaly": true/false,
    "anomaly_type": "异常类型或'none'",
    "anomaly_location": "位置描述",
    "anomaly_description": "描述",
    "multiple_events": true/false,
    "confidence": "high/medium/low"
}
若无异常，anomaly_type 设为 "none"，其余异常字段留空。"""

print("✅ 预处理函数和 Prompt 定义完成")
```

```python
import torch
from modelscope import snapshot_download
from transformers import AutoModel, AutoTokenizer, AutoConfig

model_dir = snapshot_download('OpenGVLab/InternVideo2_5_Chat_8B')
config = AutoConfig.from_pretrained(model_dir, trust_remote_code=True)
if not hasattr(config, 'llm_config'):
    config.llm_config = AutoConfig.from_pretrained(config.llm_name_or_path)

model = AutoModel.from_pretrained(
    model_dir,
    trust_remote_code=True,
    config=config,
    torch_dtype=torch.float16,
    device_map="auto",
    attn_implementation="eager"
)
tokenizer = AutoTokenizer.from_pretrained(model_dir, trust_remote_code=True)
print("✅ 模型加载完成")
```

```python
import numpy as np
import torchvision.transforms as T
from decord import VideoReader, cpu
from PIL import Image
from torchvision.transforms.functional import InterpolationMode

IMAGENET_MEAN = (0.485, 0.456, 0.406)
IMAGENET_STD  = (0.229, 0.224, 0.225)

def build_transform(input_size):
    return T.Compose([
        T.Lambda(lambda img: img.convert("RGB") if img.mode != "RGB" else img),
        T.Resize((input_size, input_size), interpolation=InterpolationMode.BICUBIC),
        T.ToTensor(),
        T.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD)
    ])

def find_closest_aspect_ratio(aspect_ratio, target_ratios, width, height, image_size):
    best_ratio_diff, best_ratio = float("inf"), (1,1)
    area = width * height
    for ratio in target_ratios:
        diff = abs(aspect_ratio - ratio[0]/ratio[1])
        if diff < best_ratio_diff:
            best_ratio_diff, best_ratio = diff, ratio
        elif diff == best_ratio_diff:
            if area > 0.5 * image_size * image_size * ratio[0] * ratio[1]:
                best_ratio = ratio
    return best_ratio

def dynamic_preprocess(image, min_num=1, max_num=6, image_size=448, use_thumbnail=False):
    ow, oh = image.size
    aspect = ow / oh
    target_ratios = sorted(
        set((i,j) for n in range(min_num, max_num+1) for i in range(1,n+1) for j in range(1,n+1) if i*j<=max_num and i*j>=min_num),
        key=lambda x: x[0]*x[1]
    )
    ta = find_closest_aspect_ratio(aspect, target_ratios, ow, oh, image_size)
    tw, th = image_size * ta[0], image_size * ta[1]
    blocks = ta[0] * ta[1]
    resized = image.resize((tw, th))
    processed = []
    for i in range(blocks):
        box = (
            (i % (tw // image_size)) * image_size,
            (i // (tw // image_size)) * image_size,
            ((i % (tw // image_size)) + 1) * image_size,
            ((i // (tw // image_size)) + 1) * image_size
        )
        processed.append(resized.crop(box))
    if use_thumbnail and len(processed) != 1:
        processed.append(image.resize((image_size, image_size)))
    return processed

def get_index(bound, fps, max_frame, first_idx=0, num_segments=32):
    if bound:
        start, end = bound[0], bound[1]
    else:
        start, end = -100000, 100000
    start_idx = max(first_idx, round(start * fps))
    end_idx = min(round(end * fps), max_frame)
    seg_size = float(end_idx - start_idx) / num_segments
    return np.array([
        int(start_idx + seg_size/2 + np.round(seg_size*idx))
        for idx in range(num_segments)
    ])

def load_video(video_path, input_size=448, max_num=1, num_segments=16):
    vr = VideoReader(video_path, ctx=cpu(0), num_threads=1)
    max_frame = len(vr) - 1
    fps = float(vr.get_avg_fps())
    pixel_values_list, num_patches_list = [], []
    transform = build_transform(input_size)
    frame_indices = get_index(None, fps, max_frame, first_idx=0, num_segments=num_segments)
    for idx in frame_indices:
        img = Image.fromarray(vr[idx].asnumpy()).convert("RGB")
        tiles = dynamic_preprocess(img, image_size=input_size, use_thumbnail=True, max_num=max_num)
        pixel_values = torch.stack([transform(tile) for tile in tiles])
        num_patches_list.append(pixel_values.shape[0])
        pixel_values_list.append(pixel_values)
    return torch.cat(pixel_values_list), num_patches_list

# 增强版 Prompt：要求模型不仅检测，还要给出推理依据、时空细节和可能的因果
ANOMALY_PROMPT = """你是一个顶尖的校园监控视频异常行为分析师。请用放大镜般的细致观看这段视频的全部帧，完成以下分析任务：

1. 异常判定：视频中是否存在任何违反校园安全规定或常规交通/行为模式的事件？
   - 交通类异常：闯红灯、逆行、超速、违停、行人闯入机动车道等。
   - 行为类异常：追逐打闹、翻越护栏、抛物、打架、攀爬围墙等。
   - 安全类异常：未戴安全帽、吸烟、违规使用手机等。

2. 如果存在异常，请提供详尽的专业分析，包含：
   - 异常类型（可以使用多标签，如 "交通违章-逆行"）
   - 异常发生的精确时空定位：具体发生在视频的哪个时间段（前段/中段/后段），描述关键帧中的场景位置（如“十字路口西北角”）。
   - 详细的行为描述：逐步描述涉事目标（人/车）的运动轨迹和交互关系，使用前后帧的逻辑关系来解释为什么判定为异常。注意说明行为的开始、持续和结束。
   - 深度推理：尝试推断异常发生的原因（如“为了赶近路而逆行”），评估事件的危险程度或对周围的影响，并指出模型判断中存在的任何不确定性。
   - 多事件标注：如果包含多个独立或关联的异常，分别列出。

3. 如果判断为正常，请说明为什么这些行为符合校园场景的常规模式，有无任何偏离常规但尚未达到异常标准的细微波动。

请严格按照以下 JSON 格式输出（不要附加其他文字）：
{
    "has_anomaly": true/false,
    "anomaly_type": "异常类型，若无则 'none'",
    "anomaly_time": "时间段描述",
    "anomaly_location": "场景位置描述",
    "anomaly_description": "逐步行为描述，包含时序逻辑",
    "reasoning": "判定依据和深度推理，包含不确定性",
    "multiple_events": true/false,
    "events": [] // 如果 multiple_events=true，在此列出每个事件的 JSON
}"""

print("✅ 预处理与深度分析 Prompt 已就绪")
```
